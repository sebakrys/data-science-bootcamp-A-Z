##Średni Błąd

```python
X = np.array([3.5, 4., 4.])
Y = np.array([2., 4., 5.])
X_mean = np.mean(X)
Y_mean = np.mean(Y)
print(f"X_mean: {X_mean}")
print(f"Y_mean: {Y_mean}")
```

```python
b_X = (1/len(X))* (abs(X-X_mean).sum())
b_Y = (1/len(Y))* (abs(Y-Y_mean).sum())
print(f"Średni bład X: {b_X}")
print(f"Średni bład Y: {b_Y}")
```

##Wariancja

```python
var_X = 1/len(X) * ((X-X_mean)**2).sum()
var_Y = 1/len(X) * ((Y-Y_mean)**2).sum()

print(f"Wariancja X: {var_X}")
print(f"Wariancja Y: {var_Y}")

std_X = np.sqrt(var_X)
std_Y = np.sqrt(var_Y)

print(f"Odchylenie standardowe X: {std_X}")
print(f"Odchylenie standardowe Y: {std_Y}")
```

```python
print(np.std(X))
print(np.std(Y))
```

##Prawdopodobieństwo

prawd. wyrzucenia parzystej liczby kostek

```python
omega = {1, 2, 3, 4, 5, 6}

A = {item for item in omega if item%2==0}
P_A = len(A) / len(omega)
print(f'ZBiór A: {A}')
print(f'Prawdopodobienstwo zdarzenia A: {P_A}')
```

prawd. wyrzucenia liczby kostek > 4

```python
omega = {1, 2, 3, 4, 5, 6}

B = {item for item in omega if item>4} # {zwróć for item in zbiór if (warunek)}
P_B = len(B) / len(omega)
print(f'ZBiór B: {B}')
print(f'Prawdopodobienstwo zdarzenia B: {P_B}')
```

###Niezależność zdarzeń

```python
intersection_AB = set.intersection(A, B)
P_intersection_AB = len(intersection_AB) / len(omega)
PA_PB = P_A *P_B
print(f'Przeciecie zbiorów A i B: {intersection_AB}\n')
print(f'Iloczyn prawdopodobienstw zdarzen A i B: {PA_PB:.4f}')
print(f'Prawdopodobienstwo iloczynu zdarzen A i B: {P_intersection_AB:.4f}\n')
check = 'Zdarzenia niezależne' if PA_PB == P_intersection_AB else 'Zdarzenia zależne'
print(check)
```

```python
C = {item for item in omega if item>2}

P_C = len(C) / len(omega)
print(f'ZBiór C: {C}')
print(f'Prawdopodobienstwo zdarzenia C: {P_C}')

intersection_AC = set.intersection(A, C)
P_intersection_AC = len(intersection_AC) / len(omega)


PA_PC = P_A *P_C


print(f'Przeciecie zbiorów A i B: {intersection_AC}\n')
print(f'Iloczyn prawdopodobienstw zdarzen A i B: {PA_PC:.4f}')
print(f'Prawdopodobienstwo iloczynu zdarzen A i B: {P_intersection_AC:.4f}\n')
check = 'Zdarzenia niezależne' if PA_PB == P_intersection_AC else 'Zdarzenia zależne'
print(check)
```
